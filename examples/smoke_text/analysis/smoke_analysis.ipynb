# Tobii Pytracker smoke output analysis
Run this notebook from the examples repository root after completing the smoke demo.


```python
from pathlib import Path
import csv

OUTPUT_ROOT = Path('output')
sessions = sorted(p for p in OUTPUT_ROOT.iterdir() if p.is_dir())
assert sessions, f'No sessions under {OUTPUT_ROOT}'
session = sessions[-1]
csv_path = session / 'data.csv'
session, csv_path
```

```python
with csv_path.open('r', encoding='utf-8', newline='') as handle:
    reader = csv.DictReader(handle, delimiter=';')
    columns = reader.fieldnames or []
    rows = list(reader)

required = {'screenshot_file','input_data','classification','user_classification','gaze_data','objects_bboxes','voice_file','voice_start_timestamp'}
assert required.issubset(columns), sorted(required.difference(columns))
assert rows, 'data.csv contains no trial rows'
assert len(rows) == 4, f'Expected 4 trial rows, found {len(rows)}'
{'rows': len(rows), 'responses': sum(bool((r.get('user_classification') or '').strip()) for r in rows), 'gaze_rows': sum(bool((r.get('gaze_data') or '').strip()) for r in rows)}
```
